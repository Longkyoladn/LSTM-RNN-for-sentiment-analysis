# so sánh với các mô hình học máy trước đây 

Logistic Regression và SVM là mô hình tuyến tính cổ điển, cần biến văn bản thành vector đặc trưng trước, thường bằng Bag-of-Words hoặc TF-IDF (có thể thêm n-gram như bigram). Chúng coi mỗi review như một “túi từ”, không quan tâm thứ tự từ. Vector đầu vào rất thưa và có số chiều lớn (hàng chục nghìn đến hàng triệu đặc trưng).

LSTM là mạng nơ-ron hồi quy. Nó nhận chuỗi từ theo đúng thứ tự (qua lớp Embedding), có bộ nhớ để giữ ngữ cảnh dài hạn, nên hiểu được những cấu trúc như phủ định (“not good”), đảo ý (“the start was boring but the ending was great”) hay sự phụ thuộc xa giữa các từ.


khi sử dụng bộ dữ liệu IMDB thì thường Logistic Regression hoặc SVM với TF-IDF và bigram thường đạt độ chính xác ngang bằng hoặc cao hơn LSTM đơn giản. Lý do là review IMDB khá dài và cảm xúc thể hiện rõ qua các từ khóa (“excellent”, “waste”, “boring”), nên thứ tự từ chỉ đóng góp thêm một phần nhỏ. Dữ liệu 25k mẫu cũng chưa đủ lớn để LSTM học Embedding từ đầu một cách vượt trội.

LSTM thắng thế khi:

+ Dùng embedding pretrained (GloVe, Word2Vec) hoặc BiLSTM/attention.
+ Review chứa nhiều châm biếm, phủ định, đảo ý phức tạp.
+ Muốn mở rộng sang bài toán khó hơn (đa lớp, văn bản dài hơn, dữ liệu lớn hơn nhiều).

SVM so với Logistic Regression: hai mô hình cho kết quả rất gần nhau với đặc trưng TF-IDF tuyến tính. SVM đôi khi nhỉnh hơn một chút về độ chính xác, còn Logistic Regression tiện hơn nhờ cho xác suất và dễ diễn giả
